```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from google.colab import files

uploaded = files.upload()

filename = list(uploaded.keys())[0]

df = pd.read_csv(filename)
```

```python
numeric_df = df.select_dtypes(include=['number'])
```

```python
correlation = numeric_df.corr()
```

```python
plt.figure(figsize=(12,8))

sns.heatmap(correlation,
            annot=True,
            cmap="coolwarm")

plt.title("Correlation Heatmap")
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.hist(df["CGPA"], bins=10)
plt.title("Histogram of CGPA")
plt.xlabel("CGPA")
plt.ylabel("Frequency")
plt.show()
```

```python
placement = df["PlacementStatus"].value_counts()

plt.figure(figsize=(6,4))
plt.bar(placement.index.astype(str), placement.values)
plt.title("Placement Status")
plt.xlabel("Placement Status")
plt.ylabel("Count")
plt.show()
```

```python
placement = df["PlacementStatus"].value_counts()

plt.figure(figsize=(6,6))
plt.pie(placement.values,
        labels=placement.index,
        autopct="%1.1f%%")

plt.title("Placement Status Percentage")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.plot(df["CGPA"])
plt.title("CGPA Line Chart")
plt.xlabel("Student Index")
plt.ylabel("CGPA")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df["CGPA"], df["CodingTestScore"])
plt.title("CGPA vs Coding Test Score")
plt.xlabel("CGPA")
plt.ylabel("Coding Test Score")
plt.show()
```

```python
plt.figure(figsize=(6,5))
plt.boxplot(df["CGPA"])
plt.title("Box Plot of CGPA")
plt.ylabel("CGPA")
plt.show()
```

```python
import seaborn as sns

plt.figure(figsize=(6,4))
sns.countplot(data=df, x="PlacementStatus")
plt.title("Placement Status Count")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df["CGPA"], kde=True)
plt.title("Distribution of CGPA")
plt.show()
```
